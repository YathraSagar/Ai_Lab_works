```python
# Helper rules based on standard conventions
def is_atom(e):     return '(' not in str(e) and type(e) is str
def is_var(e):      return is_atom(e) and e.islower()
def occurs_in(v,e): return v in str(e)

def unify(expr1, expr2):
    print(f"-> Unify({expr1}, {expr2})")

    # 1. Base cases for variables and constants
    if is_atom(expr1) or is_atom(expr2):
        if expr1 == expr2:
            return "NIL"
        elif is_var(expr1):
            return "FAILURE" if occurs_in(expr1, expr2) else [{expr2: expr1}]
        elif is_var(expr2):
            return "FAILURE" if occurs_in(expr2, expr1) else [{expr1: expr2}]
        else:
            print("  [Line 1d triggered: Different constants mismatched]")
            return "FAILURE"

    # 2 & 3. Structural checks
    if expr1[0] != expr2[0] or len(expr1) != len(expr2):
        return "FAILURE"

    # 4 & 5. Loop through arguments (skipping the predicate head at index 0)
    subst = "NIL"
    e1, e2 = list(expr1), list(expr2)

    for i in range(1, len(e1)):
        s = unify(e1[i], e2[i])

        if s == "FAILURE":
            return "FAILURE"

        if s != "NIL":
            # Apply substitutions to all remaining arguments down the line
            for j in range(i + 1, len(e1)):
                for rule in s:
                    for val, var in rule.items():
                        if e1[j] == var: e1[j] = val
                        if e2[j] == var: e2[j] = val

            subst = s if subst == "NIL" else s + subst

    # 6. Return successful unifier
    return subst

# ========================================================
# TEST RUN (Cat vs Dog)
# Represent phrases as lists: [Head, Arg1, Arg2]
# ========================================================
print("Result:", unify(["HasPet", "Tom", "Cat"], ["HasPet", "Tom", "Dog"]))
```
