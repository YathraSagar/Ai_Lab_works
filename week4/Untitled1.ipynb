
### 4-Queens Problem using Hill Climbing Search

This implementation searches for a configuration of 4 queens on a 4x4 chessboard. The state is represented as an array of 4 integers, where each index represents a column, and the value represents the row position of the queen in that column (0-indexed).

The objective function to minimize is the number of mutually attacking pairs of queens.

```python
import random

def calculate_attacks(state):
    """Calculates the number of mutually attacking pairs of queens."""
    attacks = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            # Same row or same diagonal
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                attacks += 1
    return attacks

def get_neighbors(state):
    """Generates all neighboring states by moving one queen in its column."""
    neighbors = []
    n = len(state)
    for col in range(n):
        for row in range(n):
            if state[col] != row:
                neighbor = list(state)
                neighbor[col] = row
                neighbors.append(tuple(neighbor))
    return neighbors

def print_board(state):
    """Utility to print the board layout."""
    n = len(state)
    for row in range(n):
        line = ""
        for col in range(n):
            if state[col] == row:
                line += " Q "
            else:
                line += " . "
        print(line)
    print()

def hill_climbing_4_queens(initial_state=None):
    """Performs hill climbing minimization of attacking pairs."""
    if initial_state is None:
        # Generate a random initial state
        current_state = tuple(random.randint(0, 3) for _ in range(4))
    else:
        current_state = tuple(initial_state)

    step = 0
    print(f"--- Step {step} (Initial State) ---")
    print(f"State: {current_state}")
    print_board(current_state)
    current_value = calculate_attacks(current_state)
    print(f"Attacks (Current Value): {current_value}\n")

    while True:
        neighbors = get_neighbors(current_state)
        best_neighbor = None
        best_value = current_value

        # Evaluate all neighbors
        for neighbor in neighbors:
            val = calculate_attacks(neighbor)
            if val < best_value:
                best_value = val
                best_neighbor = neighbor

        # Termination condition: Stop when no neighbor has a lower value
        if best_neighbor is None or best_value >= current_value:
            print("Termination condition met: No neighbor has a lower number of attacks than the current state.")
            print(f"Final State: {current_state} with {current_value} attacks.")
            break

        step += 1
        current_state = best_neighbor
        current_value = best_value

        print(f"--- Step {step} ---")
        print(f"Moved to State: {current_state}")
        print_board(current_state)
        print(f"Attacks (Current Value): {current_value}\n")

# Run with a defined initial state for deterministic tracking
# You can change this state to any 4-tuple of coordinates from 0 to 3
initial_state = (3, 2, 1, 0)
hill_climbing_4_queens(initial_state)
```

```python
import random
import re

def calculate_attacks(state):
    attacks = 0
    for i in range(4):
        for j in range(i + 1, 4):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                attacks += 1
    return attacks

def get_highest_valued_successor(state):
    neighbors = []
    for col in range(4):
        for row in range(4):
            if state[col] != row:
                neighbor = list(state)
                neighbor[col] = row
                neighbors.append(tuple(neighbor))

    best_neighbor = state
    min_attacks = calculate_attacks(state)

    for neighbor in neighbors:
        attacks = calculate_attacks(neighbor)
        if attacks < min_attacks:
            min_attacks = attacks
            best_neighbor = neighbor

    return best_neighbor, min_attacks

def hill_climbing_4_queens_restart():
    # Allow custom initial state input for the first run
    user_input = input("Enter initial state (e.g., '3, 2, 1, 0') or press Enter for random: ").strip()

    if user_input:
        try:
            digits = re.findall(r'\d', user_input)
            if len(digits) != 4:
                digits = [x.strip() for x in re.split(r'[\s,]+', user_input) if x.strip()]
            current_state = tuple(int(x) for x in digits)
            if len(current_state) != 4 or any(x < 0 or x > 3 for x in current_state):
                raise ValueError("State must contain 4 integers between 0 and 3.")
        except Exception as e:
            print(f"Invalid input ({e}). Generating random state.")
            current_state = tuple(random.randint(0, 3) for _ in range(4))
    else:
        current_state = tuple(random.randint(0, 3) for _ in range(4))

    restart_count = 0

    while True:
        current_attacks = calculate_attacks(current_state)
        print(f"\n--- Start Search (Run #{restart_count + 1}) ---")
        print(f"Initial state: {current_state} with {current_attacks} attacks.")

        step = 0
        local_minimum = False

        while True:
            step += 1
            neighbor, neighbor_attacks = get_highest_valued_successor(current_state)

            print(f"Step {step}: Current: {current_state} ({current_attacks} att) -> Best Neighbor: {neighbor} ({neighbor_attacks} att)")

            if neighbor_attacks >= current_attacks:
                print("Stuck in a local minimum.")
                local_minimum = True
                break

            current_state = neighbor
            current_attacks = neighbor_attacks

            if current_attacks == 0:
                print(f"\nGoal state found! {current_state} with 0 attacks.")
                return current_state

        if local_minimum:
            restart_count += 1
            print("Restarting search with a new random state...")
            current_state = tuple(random.randint(0, 3) for _ in range(4))

# Run the implementation
final_result = hill_climbing_4_queens_restart()
```
