```python
# Alpha-Beta Search Algorithm

def alpha_beta(state, alpha, beta, maximizing):

    # If leaf node, return its value
    if isinstance(state, int):
        return state

    if maximizing:
        value = float('-inf')

        for child in state:
            value = max(value, alpha_beta(child, alpha, beta, False))

            alpha = max(alpha, value)

            # Beta pruning
            if alpha >= beta:
                break

        return value

    else:
        value = float('inf')

        for child in state:
            value = min(value, alpha_beta(child, alpha, beta, True))

            beta = min(beta, value)

            # Alpha pruning
            if beta <= alpha:
                break

        return value


# Function to create game tree
def create_tree(depth, n, values):

    if depth == 0:
        return values.pop(0)

    tree = []

    for i in range(n):
        tree.append(create_tree(depth - 1, n, values))

    return tree


# Main program
print("ALPHA-BETA SEARCH ALGORITHM")

depth = int(input("Enter depth of tree: "))
n = int(input("Enter number of children per node: "))

total = n ** depth

print("Enter", total, "leaf node values:")

values = list(map(int, input().split()))

if len(values) != total:
    print("Error: Enter exactly", total, "values")

else:
    tree = create_tree(depth, n, values)

    print("\nGame Tree:", tree)

    result = alpha_beta(
        tree,
        float('-inf'),
        float('inf'),
        True
    )

    print("\nOptimal Value:", result)
```
