# Propositional Logic Solver using Python
This notebook constructs a truth table for the Knowledge Base (KB):
1. $Q \rightarrow P$
2. $P \rightarrow \neg Q$
3. $Q \lor R$

And checks the following entailments:
- Does KB $\models R$?
- Does KB $\models R \rightarrow P$?
- Does KB $\models Q \rightarrow R$?

```python
import pandas as pd
import itertools

# Define the implications and logic helper functions
def implication(a, b):
    return not a or b

# All possible truth values for P, Q, R
truth_values = [True, False]
combinations = list(itertools.product(truth_values, repeat=3))

rows = []
for P, Q, R in combinations:
    # KB sentences
    sent1 = implication(Q, P)        # Q -> P
    sent2 = implication(P, not Q)    # P -> ~Q
    sent3 = Q or R                   # Q V R

    # KB is true if all sentences are true
    kb_true = sent1 and sent2 and sent3

    # Formulas to check for entailment
    query_ii = R                     # R
    query_iii = implication(R, P)    # R -> P
    query_iv = implication(Q, R)     # Q -> R

    rows.append({
        'P': P,
        'Q': Q,
        'R': R,
        'Q -> P': sent1,
        'P -> ~Q': sent2,
        'Q v R': sent3,
        'KB (Joint)': kb_true,
        'ii) R': query_ii,
        'iii) R -> P': query_iii,
        'iv) Q -> R': query_iv
    })

df = pd.DataFrame(rows)
display(df)
```

```python
# Identify models where KB is true
models = df[df['KB (Joint)'] == True]
print("Models where KB is True:")
display(models[['P', 'Q', 'R', 'KB (Joint)']])

# Verify entailments
print("\n--- Entailment Results ---")
for query_name, col in [("ii) R", "ii) R"), ("iii) R -> P", "iii) R -> P"), ("iv) Q -> R", "iv) Q -> R")]:
    # Entailment holds if the query is True in all models where KB is True
    entails = models[col].all()
    print(f"Does KB entail {query_name}? {entails}")
```
